# OE = Odd-Even | صيغة الجبري
## Def. 0.0 - قفل الترميز - لا يتغير أبداً

| الرمز | الاتجاه | الصيغة |
|---|---|---|
| 3n+1 | نازل ↓ | C(n) إذا فردي |
| n/2 | نازل ↓ | C(n) إذا زوجي |
| 2n | صاعد ↑ | عكس n/2 - دائماً |
| (n-1)/3 | صاعد ↑ | عكس 3n+1 - بشرط (n-1) mod 3 = 0 |
| (2^k n -1)/3 | صاعد ↑ | عكس S - الحالة العامة |

**S(n) = n / 2^ν₂(n)** إذا زوجي 
**S(n) = (3n+1) / 2^ν₂(3n+1)** إذا فردي 
**S⁻¹(n) = {2n} ∪ {(2^k n -1)/3}** 

**Z + C + A = 1**

```python
"""OE - شجرة كولاتز المعكوسة الصحيحة
S(n) = (3n+1) / 2^nu2(3n+1) إذا فردي
S^-1(n) = {2n} U {(2^k n -1)/3 | 2^k n ≡ 1 (mod 3)}
"""
import matplotlib.pyplot as plt
import networkx as nx
from collections import deque

def nu2(x: int) -> int:
 """أس 2 في تحليل x."""
 k = 0
 while x % 2 == 0:
 x //= 2
 k += 1
 return k

def S(n: int) -> int:
 """خطوة كولاتز المختصرة."""
 if n % 2 == 0:
 return n // 2
 return (3 * n + 1) // (2 ** nu2(3 * n + 1))

def children_of(m: int):
 """أبناء m: كل n بحيث S(n) = m.
 - n = 2m دائماً
 - n = (m*2^k -1)/3 فردي > 1
 """
 children = [2 * m]
 for k in range(1, 40):
 num = m * (2 ** k) - 1
 if num % 3 != 0:
 continue
 n = num // 3
 if n > 1 and n % 2 == 1 and n != m:
 children.append(n)
 return children
```

```python
def build_tree(max_nodes=300, max_depth=8):
 """تبني الشبكة من الجذر 1. الحواف من الابن إلى الأب: n -> m."""
 G = nx.DiGraph()
 G.add_node(1, depth=0)
 frontier = [1]
 visited = {1}
 for depth in range(max_depth):
 nxt = []
 for m in frontier:
 for n in children_of(m):
 if n in visited:
 if not G.has_edge(n, m):
 G.add_edge(n, m)
 continue
 if len(visited) >= max_nodes:
 break
 visited.add(n)
 G.add_node(n, depth=depth+1)
 G.add_edge(n, m)
 nxt.append(n)
 frontier = nxt
 if not frontier:
 break
 return G

def verify_tree(G):
 """الشرط الوحيد الصحيح: لكل حافة n->m يجب S(n) = m."""
 ok = True
 for n, m in G.edges():
 if S(n) != m:
 print(f"⚠️ حافة غير متوافقة: {n} -> {m} , S({n})={S(n)}")
 ok = False
 if ok:
 print("✅ كل الحواف متوافقة: S(n) = m")
 return ok

def path_to_1(n: int):
 """مسار n إلى 1 بتكرار S."""
 path = [n]
 while n > 1:
 n = S(n)
 path.append(n)
 return path
```

```python
G = build_tree()
print(f"عدد العقد: {len(G.nodes())}")
print(f"العقد (مرتبة): {sorted(G.nodes())[:30]} ...")
verify_tree(G)
print()
for x in [27, 41, 31, 47, 7, 3, 5]:
 print(f"S({x}) = {S(x)} | مسار: {path_to_1(x)}")
```

```python
def draw_tree(G, title="شجرة كولاتز المعكوسة الصحيحة - OE"):
 by_depth = {}
 for node, data in G.nodes(data=True):
 by_depth.setdefault(data.get('depth', 0), []).append(node)
 pos = {}
 for d, nodes in by_depth.items():
 nodes = sorted(nodes)
 for i, node in enumerate(nodes):
 pos[node] = (i - len(nodes)/2, -d)
 plt.figure(figsize=(16, 10))
 colors = ['gold' if n==1 else 'lightblue' if n%2==0 else 'lightgreen' for n in G.nodes()]
 nx.draw(G, pos, with_labels=True, node_size=800, node_color=colors,
 font_size=8, font_weight='bold', arrows=True, arrowsize=12, edge_color='gray')
 plt.title(title, fontsize=14, fontweight='bold')
 plt.tight_layout()
 plt.savefig("oe-tree.png", dpi=200)
 plt.show()
 print("تم حفظ الصورة: oe-tree.png")

draw_tree(G)
```
